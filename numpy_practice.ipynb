```python
daily_sales = np.array([45000, 62000, 38000, 71000, 55000, 49000, 83000])
print(daily_sales)
print("Type:", type(daily_sales))
print("dtype:", daily_sales.dtype)
print("shape:", daily_sales.shape)
print("Ndim:", daily_sales.ndim)
```

```python
transportation = np.array([500, 300, 600, 800, 400, 900, 100])
inflation = 0.10

for n in transportation:
    n = (n * inflation) + n
    print (n, end="  ")

print("Type:", transportation.dtype)
print("Shape:", transportation.shape)
```
